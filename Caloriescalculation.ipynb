```python
#Calories Calculator
Calory_chart={"apple": 72,
              "banana" : 105,
              "green beans": 40,
              "carrot": 38,
              "chicken breast (3 oz roasted)": 142,
              "ground beef patty (4 oz)": 250,
              "boiled egg (1 large)": 69 ,
              "pork chop (3 oz)": 221 ,
              "rice(100g)":130 ,
              "butter(1 tablespoon)":105,
              "milk(1oz,2%)": 122,}
total_cal=0
while True:
    enter_food=input("enter food name:")
    if enter_food=="done":
        break
    if enter_food.lower() in Calory_chart:
        total_cal= total_cal+Calory_chart[enter_food.lower()]
        print("current calories:", total_cal)
    else:
        print("food not found, please try again")

print("Total calories:", total_cal)




```
