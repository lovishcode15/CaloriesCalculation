```python
#Calories Calculator
Calory_chart={"apple": 72,
              "banana" : 105,
              "green beans": 40,
              "carrot": 38,
              "chicken breast": 142,
              "ground beef patty": 250,
              "boiled egg": 69 ,
              "pork chop": 221 ,
              "rice":130 ,
              "butter":105,
              "milk": 122,}
total_cal=0
while True:
    enter_food=input("enter food name:")
    if enter_food=="done":
        break
    if enter_food.lower() in Calory_chart:
        enter_quan=int(input("enter quantity"))
        total_cal= total_cal+Calory_chart[enter_food.lower()]*enter_quan
        print("current calories:", total_cal)
    else:
        print("food not found, please try again")

print("Total calories:", total_cal)




```
